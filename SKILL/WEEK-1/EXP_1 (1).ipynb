```python
import os
from google.colab import userdata

try:
    # Retrieve the Gemini API key from Colab Secrets securely
    api_key = userdata.get('GEMINI_API_KEY')
    os.environ["GEMINI_API_KEY"] = api_key
    print("API key loaded successfully!")
except userdata.SecretNotFoundError:
    print("ERROR: 'GEMINI_API_KEY' secret not found. Please click the key icon in the left sidebar, add a secret named 'GEMINI_API_KEY' with your actual key, and grant this notebook access to it.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
import time
from google import genai

client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])
model = "gemini-3.5-flash-lite"

question = input("Ask your question: ")      # type: What is AI?

response = None
for attempt in range(1, 5):
    try:
        start = time.time()
        response = client.models.generate_content(model=model, contents=question)
        latency = (time.time() - start) * 1000
        break
    except Exception as e:
        print(f"Attempt {attempt} failed: {e}. Retrying...")
        if attempt < 4:
            time.sleep(attempt * 2)

if response is not None:
    usage = response.usage_metadata
    print("\nQuestion:", question)
    print("Answer  :", response.text)
    print("\nInput tokens :", usage.prompt_token_count)
    print("Output tokens:", usage.candidates_token_count)
    print("Total tokens :", usage.total_token_count)
    print("Latency (ms) :", round(latency))
else:
    print("\nAll attempts to reach the Gemini API failed. Please make sure your 'GEMINI_API_KEY' is valid and active.")
```
